```python
import re
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, LancasterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

text = """
Badminton is my favorite sport because it keeps me active and healthy.
I enjoy playing badminton with my friends in my free time.
It improves my concentration, speed, and physical fitness.
Playing this sport teaches me teamwork, discipline, and patience.
I feel happy and refreshed whenever I play badminton.
"""

print("\nOriginal Text:\n", text)
cleaned_text = text.lower()
cleaned_text = cleaned_text.translate(
    str.maketrans('', '', string.punctuation)
)

print("\nLowercase Text Without Punctuation:\n", cleaned_text)

words = word_tokenize(cleaned_text)
sentences = sent_tokenize(text)

print("\nWord Tokens:\n", words)
print("\nSentence Tokens:\n", sentences)

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

print("\nWords After Stopword Removal:\n", filtered_words)

frequency = nltk.FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
for word, count in frequency.most_common():
    print(f"{word}: {count}")
```

```python
porter = PorterStemmer()
lancaster = LancasterStemmer()

porter_results = [porter.stem(word) for word in filtered_words]
lancaster_results = [lancaster.stem(word) for word in filtered_words]

lemmatizer = WordNetLemmatizer()

lemma_results = [
    lemmatizer.lemmatize(word) for word in filtered_words
]

print("\nOriginal Word | Porter Stemmer | Lancaster Stemmer | Lemmatizer")
print("-" * 70)

for original, p, l, lemma in zip(
    filtered_words,
    porter_results,
    lancaster_results,
    lemma_results
):
    print(f"{original:14} | {p:14} | {l:17} | {lemma}")
```

```python
long_words = re.findall(r'\b[A-Za-z]{6,}\b', text)

print("\nWords With More Than 5 Letters:\n", long_words)

numbers = re.findall(r'\b\d+(?:\.\d+)?\b', text)

print("\nNumbers Found:\n", numbers)

capitalized_words = re.findall(r'\b[A-Z][a-zA-Z]*\b', text)

print("\nCapitalized Words:\n", capitalized_words)

alphabet_words = re.findall(r'\b[A-Za-z]+\b', text)

print("\nAlphabet-Only Words:\n", alphabet_words)

vowel_words = re.findall(r'\b[AEIOUaeiou][A-Za-z]*\b', text)

print("\nWords Starting With a Vowel:\n", vowel_words)
```

```python

import re

def custom_tokenize(text):
    pattern = r"\d+\.\d+|[A-Za-z]+(?:[-'][A-Za-z]+)+|[A-Za-z]+|\d+"
    tokens = re.findall(pattern, text)
    return tokens

text = """
Badminton is my favorite sport because it keeps me active and healthy.
I enjoy playing badminton with my friends in my free time.
It improves my concentration, speed, and physical fitness.
Playing this sport teaches me teamwork, discipline, and patience.
I feel happy and refreshed whenever I play badminton.
"""

tokens = custom_tokenize(text)

print("Original Text:\n", text)
print("\nCustom Tokens:\n", tokens)

sample_text = "I love state-of-the-art technology. It isn't difficult. Value is 3.14 in 2026!"

print("\nSample Text:\n", sample_text)
print("\nSample Tokens:\n", custom_tokenize(sample_text))
```
