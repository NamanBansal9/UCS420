```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print("original array:", arr)

add_arr = arr + 2
print("after addition of 2:", add_arr)
mul_arr = arr * 3
print("after multiplying by 3:", mul_arr)
div_arr = arr / 2
print("after dividing by 2:", div_arr)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("original array:", arr)
print("reversed array:", reversed_arr)
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_freq_x = values[np.argmax(counts)]
indices_x = np.where(x == most_freq_x)[0]
print("array x:", x)
print("frequent value:", most_freq_x)
print("indices:", indices_x)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
most_freq_y = values[np.argmax(counts)]
indices_y = np.where(y == most_freq_y)[0]
print("array y:", y)
print("most frequent value:", most_freq_y)
print("indices:", indices_y)
```

```python
arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("2-D array:\n", arr)

element1 = arr[0, 1]
print("1st Row, 2nd column element:", element1)

element2 = arr[2, 0]
print("3rd Row, 1st column element:", element2)
```

```python
naman = np.linspace(10, 100, 25)
print("Array naman:", naman)

print("Dimensions:", naman.ndim)
print("Shape:", naman.shape)
print("Total Elements:", naman.size)
print("Data Type:", naman.dtype)
print("Total Bytes:", naman.nbytes)

transpose_reshape = naman.reshape(25, 1)
print("Transpose using reshape():\n", transpose_reshape)
```

Since `naman` is a 1-D array, applying the `.T` attribute does not change it, as transpose has no effect on a 1-D array. It only works properly on arrays with 2 or more dimensions. To actually see a transpose-like effect using `.T`, the array first needs to be reshaped into a 2-D array.

```python
print("using T attribute on 1-D array (no change):", naman.T.shape)
naman_2 = naman.reshape(25, 1)
print("using T attribute on reshaped 2-D array:\n", naman_2.T)
```

```python
ucs420_naman = np.array([[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])
print("Array ucs420_naman:\n", ucs420_naman)

print("mean:", np.mean(ucs420_naman))
print("median:", np.median(ucs420_naman))
print("max:", np.max(ucs420_naman))
print("Mmn:", np.min(ucs420_naman))
print("Uuique elements:", np.unique(ucs420_naman))
```

```python
reshaped_ucs420_naman = ucs420_naman.reshape(4, 3)
print("reshaped Array (4x3):\n", reshaped_ucs420_naman)
```

```python
resized_ucs420_naman = np.resize(ucs420_naman, (2, 3))
print("resized Array (2x3):\n", resized_ucs420_naman)
```
