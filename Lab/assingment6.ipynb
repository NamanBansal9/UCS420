Q1. Consider the following sensor readings: 
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40]) 
A sensor has an average measurement error of +2°C. 
Perform the following without using a for loop: 
a. Add 2°C to every reading using vectorization. 
b. Convert all temperatures from Celsius to Fahrenheit.
c. Identify all readings greater than 32°C using Boolean indexing. 
d. Count how many readings exceed 32°C. 
e. Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value. 

```python
import numpy as np

temperature = np.array([25,28,31,35,38,27,33,40])
print("original array:", temperature)

cor_temp = temperature + 2
print("after addition of 2:", cor_temp)
```

```python
mul = (9/5)*cor_temp
farh = mul + 32
print("temperature in fahrenheit",farh)
```

```python
print(cor_temp > 32)
```

```python
ct = np.sum(cor_temp > 32)
ct
```

Vectorization allows NumPy to perform an operation on an entire array at once instead of using a for loop for every element. It is generally faster because NumPy performs many operations using optimized low-level implementations.

Boolean indexing creates a Boolean condition such as:
cor_temp > 32
which can directly select only the elements satisfying the condition.

Therefore, vectorization and Boolean indexing make code:
Faster
Shorter
Easier to read
More efficient for large datasets
Less dependent on explicit Python loops


Q2. The following matrix represents the number of daily steps recorded by four users 
over three days: 
        steps = np.array([ 
        [5000, 6200, 7100], 
        [8000, 7500, 9000], 
        [4500, 5100, 4800], 
        [9000, 8500, 9500]]) 
Perform the following: 
a. Calculate the total steps recorded. 
b. Calculate the mean number of steps. 
c. Find the maximum and minimum values. 
d. Calculate the total steps for each day using axis=0. 
e. Calculate the total steps for each user using axis=1. 
f. Find the user/day position containing the maximum number of steps using argmax().

```python
steps = np.array([[5000,6200,7100],[8000,7500,9000],[4500,5100,4800],[9000,8500,9500]])
total_steps = np.sum(steps)
total_steps
```

```python
mn = np.mean(steps)
mn
```

```python
mx = np.max(steps)
mn = np.min(steps)
print(mx)
print(mn)
```

```python
total_steps_col = np.sum(steps,axis= 0)
total_steps_row = np.sum(steps,axis= 1)
print(total_steps_col)
print(total_steps_row)
```

```python
max_position = np.argmax(steps)
max_position_2d = np.unravel_index(np.argmax(steps), steps.shape)

print("f. Flattened position of maximum:", max_position)
print("   User/day position of maximum:", max_position_2d)
```

Q4. A simple cognitive-assistive system wants to predict a user's assistance score based 
on three factors: 
• Sleep hours  
• Activity level  
• Stress level  
Consider:
            X = np.array([ 
            [6, 70, 3], 
            [5, 50, 6], 
            [8, 80, 2], 
            [4, 30, 8] 
            ]) 

            y = np.array([40, 65, 30, 85]) 

Here, each row represents one user and each column represents a feature. 
Perform the following: 
a. Display the shape and dimensions of X. 
b. Calculate: X.T and explain what the transpose represents. 
c. Calculate the matrix product: X.T @ X 
d. Calculate the matrix inverse using: np.linalg.inv() where appropriate. 
e. Implement the Ordinary Least Squares equation: 𝛽 = (𝑋𝑋) 𝑋𝑦 using NumPy. 
f. Explain what the resulting coefficients represent in the context of the three user 
features. 
g. A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to obtain a predicted assistance score.

```python
x = np.array([
    [6,70,3],
    [5,50,6],
    [8,80,2],
    [4,30,8]
])
y = np.array([40,65,30,85])
```

```python
print(x.shape)
print(x.ndim)
```

```python
xtrans = x.T
print(xtrans)
```

```python
xtx = x.T @ x
print(xtx)
```

```python
xtx_inverse = np.linalg.inv(xtx)
print(xtx_inverse)
```

```python
bi = xtx_inverse @ xtrans @ y
print(bi)
```

```python
print("Coefficient interpretation:")
print("Sleep coefficient:", bi[0])
print("Activity coefficient:", bi[1])
print("Stress coefficient:", bi[2])
```

```python
new_user = np.array([5, 40, 7])

predicted_score = new_user @ bi

print("\ng. Predicted assistance score:", predicted_score)
```
